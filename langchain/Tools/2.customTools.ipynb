```python
print("using tool decorator")
```

```python
from langchain_core.tools import tool
```

```python
# Step 1 - create a function

def multiply(a, b):
    """Multiply two numbers"""
    return a*b
```

```python
# Step 2 - add type hints

def multiply(a: int, b:int) -> int:  # add variable types and return type is actually called as hint....!
    """Multiply two numbers"""
    return a*b
```

```python
# Step 3 - add tool decorator

@tool
def multiply(a: int, b:int) -> int:
    """Multiply two numbers"""
    return a*b
```

```python
result = multiply.invoke({"a":3, "b":5})
```

```python
print(result)
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
```

```python
print(multiply.args_schema.model_json_schema())
```

```python
from langchain_core.tools import StructuredTool
from langchain_core.output_parsers import PydanticOutputParser
from pydantic import BaseModel, Field
from typing import Literal
```

```python
# Step 1 : create a function

def calculate(a: int, b: int) -> int:
    return a * b;
```

```python
class ArgsSchema(BaseModel):
    a: int= Field(description="first input")
    b: int = Field(description="Second Input")
```

```python
tool = StructuredTool.from_function(
    func=calculate,
    description="Use this function if you want to multiply 2 numbers",
    name='CalculatorTool',
    args_schema=ArgsSchema
)
```

```python
tool.invoke({"a": 2, "b": 3})
```

```python
print(tool.name)
print(tool.description)
print(tool.args)
```

```python
print("3.) BaseClass Tool")
```

```python
from langchain.tools import BaseTool
from typing import Type
```

```python
# arg schema using pydantic

class MultiplyInput(BaseModel):
    a: int = Field(required=True, description="The first number to add")
    b: int = Field(required=True, description="The second number to add")
```

```python
class MultiplyTool(BaseTool):
    name: str = "multiply"
    description: str = "Multiply two numbers"

    args_schema: Type[BaseModel] = MultiplyInput

    def _run(self, a: int, b: int) -> int:
        return a * b
```

```python
multiply_tool = MultiplyTool()
```

```python
result = multiply_tool.invoke({'a':3, 'b':3})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)

print(multiply_tool.args)
```
