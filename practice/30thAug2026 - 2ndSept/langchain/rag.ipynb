```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

from langchain_openai import OpenAIEmbeddings
from dotenv import load_dotenv

from langchain_community.vectorstores import Chroma
from langchain_core.documents import Document
```

```python
# loading the doc
# splitting the doc into chunks
# storing it in the vector store
# creating the retriever
# query flow

load_dotenv()
```

```python
pyPDFLoader = PyPDFLoader("./dl-curriculum.pdf")
textSplitter = RecursiveCharacterTextSplitter(chunk_size=300, chunk_overlap=100, separators=["\n\n", "\n", " ", ""])

embeddingServer = OpenAIEmbeddings()
store = Chroma(embedding_function=embeddingServer)
```

```python
documents = pyPDFLoader.load()
```

```python
for document in documents:
    content = textSplitter.split_text(document.page_content)
    if not content:
        continue
    store.add_documents([
        Document(page_content=c, metadata=document.metadata) for c in content
    ])
    
```

```python
store.similarity_search("lstm?", k=3)
```

```python
retriever = store.as_retriever(search_type ="mmr",search_kwargs={"k": 3})
```

```python
retrieved_docs = retriever.invoke("lstm?")
retrieved_docs
```
