```python
from langchain_community.vectorstores import Chroma
from langchain_core.documents import Document

# embedding model 
from langchain_openai import OpenAIEmbeddings

# tools
from langchain_core.tools import tool

from dotenv import load_dotenv
```

```python

load_dotenv()
embeddingModel = OpenAIEmbeddings();
```

```python
datastore = Chroma(collection_name='test', embedding_function=embeddingModel)
```

```python
document1 = Document(page_content="Virat kohli is a good guy", metadata={"topic":"cricket"})
document2 = Document(page_content="ronaldo is a footballer", metadata={"topic":"soccer"})
```

```python
datastore.add_documents([document1, document2])
```

```python
response = datastore.similarity_search_with_score(query = 'Who is the cricketer', k = 1, filter={"topic": "cricket"})
print(response)
```

```python
mmrRetriever = datastore.as_retriever(search_type="mmr", search_kwargs={"k": 1})
```
