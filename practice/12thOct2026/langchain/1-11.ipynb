### OUTPARSERS & STRUCTURED OUTPUT

#### STRUCTURED OUTPUT

```python
from pydantic import BaseModel, Field
from typing import Annotated, Literal, TypedDict
```

```python
# Pydantic

class ReviewPydantic(BaseModel):
    review: str = Field(..., description="This field is required")
    rating: int = Field(..., gt=0, le=5,   description = "Rating")
    isItGood: Literal['good', 'bad'] = Field(...,  description="is it good")

class ReviewTypedDict(TypedDict):
    review: Annotated[str,"Review"]
    rating: Annotated[int, "Rating"]
    isItGood: Annotated[Literal['good', 'bad'], 'is it good...!']
```

```python
modelWithPydanticStructuedOutput = chatModel.with_structured_output(ReviewPydantic)
```

```python
pydanticResponse = modelWithPydanticStructuedOutput.invoke("Review of Avatar")
```

```python
pydanticResponse
```

```python
pydanticResponse.dict()['review']
```

```python
modelWithTypedStructuredOutput = chatModel.with_structured_output(ReviewTypedDict)
```

```python
typedDictOutput = modelWithTypedStructuredOutput.invoke("Review of Avatar")
```

```python
typedDictOutput['review']
```

#### Pydantic Object

```python

from langchain_core.output_parsers import PydanticOutputParser, StrOutputParser
```

```python
pydanticOutputParser = PydanticOutputParser(pydantic_object=ReviewPydantic)
stroutputParser = StrOutputParser()
```

```python

promptTemplateForPydanticObject = PromptTemplate(
template="""'{topic}' \n  '{format_instuctions}'""",
input_variables=['topic'],
validate_template=True,
partial_variables={"format_instuctions": pydanticOutputParser.get_format_instructions()}
)
```

```python
chain4=  promptTemplateForPydanticObject | chatModel | pydanticOutputParser
chain4.invoke({"topic": "reivew of avatar"})
```

### Chains


```python
# RunnableSequence, RunnableParallel , RunnableBranch, RunnableLambda, RunnablePassThrough
```

```python
# Runnable Parallel
from langchain_core.runnables import RunnableParallel, RunnableSequence, RunnablePassthrough, RunnableBranch, RunnableLambda
```

```python
# 1 chain gets the Review, other chain gets the rating , and finally a sequential chain summarize both response of both the chains
```

```python
parallelChain = RunnableParallel({
    "review": RunnableLambda(lambda x: f"Review of the film {x['movie']}") | chatModel | stroutputParser,
    "rating": RunnableLambda(lambda x: f"Rating of the film {x['movie']}") | chatModel | stroutputParser,
    "movie": RunnableLambda(lambda x: x["movie"])
})
```

```python
parllelChainPrompt = PromptTemplate(template="Summarize this {review} and {rating} of {movie}", input_variables=['review', 'rating', 'movie'], validate_template=True)

chain5 = parallelChain | parllelChainPrompt | chatModel | stroutputParser
```

```python
chain5.invoke({"movie": 'titanic'})
```

```python
# Runnable Branch
import json
# get the movie rating and based on that rating, print a message

runnableBranch = RunnableBranch(
    (lambda x: x['rating'] > 2, RunnableLambda(lambda x: "Thanks for positive review") | chatModel | stroutputParser),
    (lambda x: x['rating'] < 2, RunnableLambda(lambda x: "Please give feedback") | chatModel | stroutputParser),
    RunnableLambda(lambda x: "Who are you") | chatModel | stroutputParser)

chain4=  promptTemplateForPydanticObject | chatModel | pydanticOutputParser | RunnableLambda(lambda x: json.loads(x.json())) | runnableBranch
chain4.invoke({"topic": "reivew of avatar"})
```
