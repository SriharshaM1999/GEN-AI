### Text Splitters

```python
from langchain_text_splitters import CharacterTextSplitter, RecursiveCharacterTextSplitter
from langchain_experimental.text_splitter import SemanticChunker
```

```python
recursiveTextSplitter = RecursiveCharacterTextSplitter(chunk_size = 10, chunk_overlap = 5)
recursiveTextSplitterTikToken = RecursiveCharacterTextSplitter.from_tiktoken_encoder(chunk_size = 11, chunk_overlap = 5, encoding_name="cl100k_base")
```

```python
recursiveTextSplitter.split_documents([textDoc])
```

```python
recursiveTextSplitterTikToken.split_documents([textDoc])
```

```python
from langchain_experimental.text_splitter import SemanticChunker
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

semanticChunker = SemanticChunker(
    embeddings,
    breakpoint_threshold_type="percentile",   # split at the biggest 5% meaning changes
    number_of_chunks=100
)
```

```python
semanticChunker.split_documents([textDoc])
```

### Vector stores and retrievers

```python
from langchain_community.vectorstores import FAISS, Chroma
from langchain_openai import OpenAIEmbeddings
```

```python
store = Chroma(embedding_function= OpenAIEmbeddings(), persist_directory='./practice', collection_name='practice')
```

```python
store.add_documents([textDoc, textDoc], ids = ["1", "2"])
```

```python
store.update_documents(['1','2'], [textDoc, textDoc])
```

```python
store.similarity_search_with_relevance_scores("At dawn", k = 3)
store.max_marginal_relevance_search("At dawn", k = 3, fetch_k=5, lambda_mult=0.2)
```

```python
store.similarity_search("", k = 3, filter={'source': './content.txt'})
```

```python
store.similarity_search("", k = 3, where_document = {"$contains": "copper"})
```

```python
# retrievers
 # 3 types
  # similarity
  # mmr
  # re-ranking
  #  
```

```python
retriever = store.as_retriever(search_type = 'similarity', search_kwargs={"k": 2})
mmrRetreiver = store.as_retriever(search_type = 'mmr', search_kwargs={"k": 2, "fetch_k": 5, "lambda_mul": 0.2})
```

```python
mmrRetreiver.invoke("banana")
```
